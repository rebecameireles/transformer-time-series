```python
#2
import math
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
import optuna

# ---------- Configurações gerais ----------
SEED = 42

LEN_ENTRADA = 64     # passos do passado analisados pelo encoder
LEN_SAIDA = 12       # passos futuros 
LEN_CONTEXT = 12     # contexto dinâmico recente entregue ao decoder

N_TRIALS = 20        # número de tentativas do Optuna
EPOCAS_OPTUNA = 10   # épocas reduzidas por trial para poupar tempo de GPU
EPOCAS_FINAL = 40    # épocas do treino final com a melhor configuração

# Fixando sementes para reprodutibilidade
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

optuna.logging.set_verbosity(optuna.logging.WARNING)

# Detecta a GPU do VLab
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Rodando em:", DEVICE)
```

```python
#3
# ---------- 1) Criar série temporal sintética multivariada ----------
n_passos = 2000
t = np.arange(n_passos)
ruido = lambda: np.random.normal(0, 0.05, n_passos)

df = pd.DataFrame({
    "seno": np.sin(2 * np.pi * t / 50) + ruido(),
    "cosseno": np.cos(2 * np.pi * t / 80) + ruido(),
    "onda_mista": (
        0.5 * np.sin(2 * np.pi * t / 25)
        + 0.5 * np.sin(2 * np.pi * t / 100)
        + ruido()
    ),
})

print(df.head())
print(df.describe())

df.iloc[:300].plot(
    figsize=(10, 3),
    title="Série sintética (primeiros 300 passos)"
)
plt.show()

# ---------- 2) Separação temporal cronológica (sem embaralhar) ----------
n = len(df)
corte_treino = int(n * 0.70)
corte_val = int(n * 0.85)

treino_df = df.iloc[:corte_treino]
val_df = df.iloc[corte_treino:corte_val]
teste_df = df.iloc[corte_val:]

print("\nDivisão dos dados:")
print("Treino:", len(treino_df))
print("Validação:", len(val_df))
print("Teste:", len(teste_df))

# ---------- 3) Normalização (baseada apenas no treino para evitar vazamento) ----------
media = treino_df.mean()
desvio = treino_df.std()

treino = ((treino_df - media) / desvio).values.astype(np.float32)
val = ((val_df - media) / desvio).values.astype(np.float32)
teste = ((teste_df - media) / desvio).values.astype(np.float32)

N_FEAT = treino.shape[1]

# ---------- 4) Criar janelas deslizantes (passado -> futuro) ----------
def criar_janelas(dados, len_in, len_out):
    src, tgt = [], []
    for i in range(len(dados) - len_in - len_out + 1):
        src.append(dados[i : i + len_in])
        tgt.append(dados[i + len_in : i + len_in + len_out])
    return torch.tensor(np.stack(src)), torch.tensor(np.stack(tgt))

train_src, train_tgt = criar_janelas(treino, LEN_ENTRADA, LEN_SAIDA)
val_src, val_tgt = criar_janelas(val, LEN_ENTRADA, LEN_SAIDA)
teste_src, teste_tgt = criar_janelas(teste, LEN_ENTRADA, LEN_SAIDA)

print("\nDimensões:")
print("Treino:", train_src.shape, train_tgt.shape)
print("Validação:", val_src.shape, val_tgt.shape)
print("Teste:", teste_src.shape, teste_tgt.shape)

# Função auxiliar para criar DataLoaders
def make_loader(src, tgt, batch_size, shuffle):
    return DataLoader(TensorDataset(src, tgt), batch_size=batch_size, shuffle=shuffle)
```

```python
#4
# ---------- Codificação Posicional senoidal ----------
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(max_len).unsqueeze(1).float()
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        self.register_buffer("pe", pe.unsqueeze(0))

    def forward(self, x):
        return x + self.pe[:, : x.size(1)]

# Rede Feed-Forward interna
def feed_forward(d_model, d_ff, dropout):
    return nn.Sequential(
        nn.Linear(d_model, d_ff),
        nn.GELU(),
        nn.Dropout(dropout),
        nn.Linear(d_ff, d_model)
    )

# ---------- Camada do Encoder com arquitetura Pre-LN ----------
class EncoderLayer(nn.Module):
    def __init__(self, d_model, nhead, d_ff, dropout):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, nhead, dropout=dropout, batch_first=True)
        self.ff = feed_forward(d_model, d_ff, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.drop = nn.Dropout(dropout)

    def forward(self, x):
        # Pre-LN: normaliza antes da autoatenção
        x_norm = self.norm1(x)
        attn_out, _ = self.attn(x_norm, x_norm, x_norm, need_weights=False)
        x = x + self.drop(attn_out)
        
        # Pre-LN: normaliza antes do feed-forward
        x = x + self.drop(self.ff(self.norm2(x)))
        return x

# ---------- Camada do Decoder com arquitetura Pre-LN ----------
class DecoderLayer(nn.Module):
    def __init__(self, d_model, nhead, d_ff, dropout):
        super().__init__()
        self.self_attn = nn.MultiheadAttention(d_model, nhead, dropout=dropout, batch_first=True)
        self.cross_attn = nn.MultiheadAttention(d_model, nhead, dropout=dropout, batch_first=True)
        self.ff = feed_forward(d_model, d_ff, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.norm3 = nn.LayerNorm(d_model)
        self.drop = nn.Dropout(dropout)

    def forward(self, y, memoria, mascara_causal):
        # 1) Autoatenção causal mascarada com Pre-LN
        y_norm1 = self.norm1(y)
        s_attn, _ = self.self_attn(y_norm1, y_norm1, y_norm1, attn_mask=mascara_causal, need_weights=False)
        y = y + self.drop(s_attn)
        
        # 2) Atenção cruzada com a memória do encoder com Pre-LN
        y_norm2 = self.norm2(y)
        c_attn, _ = self.cross_attn(y_norm2, memoria, memoria, need_weights=False)
        y = y + self.drop(c_attn)
        
        # 3) Feed-forward com Pre-LN
        y = y + self.drop(self.ff(self.norm3(y)))
        return y

# ---------- Transformer Completo ----------
class TransformerSerieTemporal(nn.Module):
    def __init__(self, n_feat, d_model=64, nhead=4, n_layers=2, dropout=0.1, len_context=12):
        super().__init__()
        self.d_model = d_model
        self.len_context = len_context
        d_ff = 4 * d_model
        
        self.entrada_enc = nn.Linear(n_feat, d_model)
        self.entrada_dec = nn.Linear(n_feat, d_model)
        self.pos = PositionalEncoding(d_model)
        self.drop = nn.Dropout(dropout)
        
        self.encoder = nn.ModuleList([EncoderLayer(d_model, nhead, d_ff, dropout) for _ in range(n_layers)])
        self.decoder = nn.ModuleList([DecoderLayer(d_model, nhead, d_ff, dropout) for _ in range(n_layers)])
        
        # Camada de normalização final necessária no Pre-LN antes da saída linear
        self.norm_final = nn.LayerNorm(d_model)
        self.saida = nn.Linear(d_model, n_feat)

        # Inicialização explícita dos pesos (Xavier Uniform) para acelerar a convergência
        self._init_pesos()

    def _init_pesos(self):
        for p in self.parameters():
            if p.dim() > 1:
                nn.init.xavier_uniform_(p)

    def codificar(self, src):
        x = self.drop(self.pos(self.entrada_enc(src) * math.sqrt(self.d_model)))
        for camada in self.encoder:
            x = camada(x)
        return x

    def decodificar(self, dec_in, memoria):
        T = dec_in.size(1)
        mascara = torch.triu(torch.ones(T, T, dtype=torch.bool, device=dec_in.device), diagonal=1)
        y = self.drop(self.pos(self.entrada_dec(dec_in) * math.sqrt(self.d_model)))
        for camada in self.decoder:
            y = camada(y, memoria, mascara)
        return self.saida(self.norm_final(y))

    def forward(self, src, dec_in):
        memoria = self.codificar(src)
        return self.decodificar(dec_in, memoria)

    @torch.no_grad()
    def prever(self, src, passos=LEN_SAIDA):
        self.eval()
        memoria = self.codificar(src)
        
        # Inicia a entrada com o contexto recente + último ponto conhecido do passado
        contexto = src[:, -self.len_context:, :].clone()
        ultimo_ponto = src[:, -1:, :]
        dec_in = torch.cat([contexto, ultimo_ponto], dim=1)
        
        # Realimenta ponto a ponto recursivamente
        for _ in range(passos):
            saida = self.decodificar(dec_in, memoria)
            proximo_ponto = saida[:, -1:, :]
            dec_in = torch.cat([dec_in, proximo_ponto], dim=1)
            
        return dec_in[:, -passos:, :]
```

```python
#5
LEN_CONTEXT = 12

# Prepara a entrada do decoder no treino com Teacher Forcing
def montar_entrada_dec(src, tgt):
    contexto = src[:, -LEN_CONTEXT:, :]
    futuro_in = torch.cat([src[:, -1:, :], tgt[:, :-1, :]], dim=1)
    return torch.cat([contexto, futuro_in], dim=1)

# Avalia o modelo sem gabarito futuro (loss autorregressiva real)
@torch.no_grad()
def avaliar_autoregressivo(modelo, src, tgt):
    modelo.eval()
    loader = make_loader(src, tgt, 256, shuffle=False)
    total, n = 0.0, 0
    for src_b, tgt_b in loader:
        src_b = src_b.to(DEVICE)
        tgt_b = tgt_b.to(DEVICE)
        pred = modelo.prever(src_b, passos=LEN_SAIDA)
        perda = nn.functional.mse_loss(pred, tgt_b, reduction="sum")
        total += perda.item()
        n += tgt_b.numel()
    return total / n

# Loop de treinamento com clip de gradientes e Early Stopping
def treinar(params, epocas, verbose=False, trial=None):
    modelo = TransformerSerieTemporal(
        N_FEAT,
        params["d_model"],
        params["nhead"],
        params["n_layers"],
        params["dropout"],
        len_context=LEN_CONTEXT
    ).to(DEVICE)

    otimizador = torch.optim.AdamW(
        modelo.parameters(),
        lr=params["lr"],
        weight_decay=1e-4
    )
    criterio = nn.MSELoss()

    loader_tr = make_loader(train_src, train_tgt, params.get("batch_size", 64), shuffle=True)
    loader_val = make_loader(val_src, val_tgt, 256, shuffle=False)

    paciencia = 6 if trial is not None else 10
    melhor_loss_auto = float("inf")
    melhor_estado = None
    melhor_epoca = 0
    val_loss = float("nan")

    for epoca in range(1, epocas + 1):
        modelo.train()
        for i_batch, (src, tgt) in enumerate(loader_tr):
            src, tgt = src.to(DEVICE), tgt.to(DEVICE)
            dec_in = montar_entrada_dec(src, tgt)
            saida = modelo(src, dec_in)

            perda = criterio(saida[:, -LEN_SAIDA:, :], tgt)
            otimizador.zero_grad()
            perda.backward()

            # Checagem leve de gradiente (pruning em custo zero se o gradiente colapsar)
            if epoca == 1 and i_batch == 0 and trial is not None:
                g_norm = sum(p.grad.norm()**2 for p in modelo.parameters() if p.grad is not None)
                if torch.isnan(g_norm) or g_norm.item() < 1e-7:
                    raise optuna.TrialPruned()

            nn.utils.clip_grad_norm_(modelo.parameters(), 1.0)
            otimizador.step()

        # Validação com Teacher Forcing
        modelo.eval()
        total, n = 0.0, 0
        with torch.no_grad():
            for src, tgt in loader_val:
                src, tgt = src.to(DEVICE), tgt.to(DEVICE)
                dec_in = montar_entrada_dec(src, tgt)
                saida = modelo(src, dec_in)
                p = criterio(saida[:, -LEN_SAIDA:, :], tgt)
                total += p.item() * len(src)
                n += len(src)
        val_loss = total / n

        # Validação real autorregressiva
        val_auto = avaliar_autoregressivo(modelo, val_src, val_tgt)

        # Salva o melhor checkpoint com base na avaliação autorregressiva
        if val_auto < melhor_loss_auto - 1e-4:
            melhor_loss_auto = val_auto
            melhor_epoca = epoca
            melhor_estado = {k: v.detach().cpu().clone() for k, v in modelo.state_dict().items()}

        if verbose and (epoca % 5 == 0 or epoca == epocas):
            print(
                f"Época {epoca:02d}/{epocas} | "
                f"Teacher forcing: {val_loss:.5f} | "
                f"Autorregressiva: {val_auto:.5f} | "
                f"Melhor: {melhor_loss_auto:.5f}"
            )

        # Early Stopping se não melhorar mais
        if epoca - melhor_epoca >= paciencia:
            if verbose:
                print(f"\nEarly stopping na época {epoca}. Melhor época: {melhor_epoca}")
            break

    # Restaura o melhor modelo encontrado
    if melhor_estado is not None:
        modelo.load_state_dict(melhor_estado)
    modelo.melhor_epoca = melhor_epoca
    modelo.melhor_loss_auto = melhor_loss_auto

    if verbose:
        print(f"\nModelo recuperado da época {melhor_epoca} | Loss autorregressiva: {melhor_loss_auto:.5f}")

    return modelo, val_loss
```

```python
#6
# Função objetivo para busca de hiperparâmetros leves no Optuna
def objetivo(trial):
    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)

    params = {
        "lr": trial.suggest_float("lr", 1e-4, 5e-3, log=True),
        "d_model": trial.suggest_categorical("d_model", [32, 64, 128]),
        "nhead": trial.suggest_categorical("nhead", [2, 4]),
        "n_layers": trial.suggest_int("n_layers", 2, 4), # faixas leves de camadas
        "dropout": trial.suggest_float("dropout", 0.0, 0.2),
    }

    modelo, _ = treinar(params, epocas=EPOCAS_OPTUNA, trial=trial)
    return modelo.melhor_loss_auto

estudo = optuna.create_study(
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=SEED)
)
estudo.optimize(objetivo, n_trials=N_TRIALS)

# Exibe o resumo de todos os trials testados
print(
    estudo.trials_dataframe()[
        [
            "number", "value", "params_lr",
            "params_d_model", "params_nhead",
            "params_n_layers", "params_dropout"
        ]
    ]
)

print(f"\nMelhor trial: #{estudo.best_trial.number} | loss autorregressiva = {estudo.best_value:.5f}")
print("Melhores hiperparâmetros:", estudo.best_params)
```

```python
#7
# Treinamento final com os melhores parâmetros encontrados
modelo, val_loss = treinar(
    estudo.best_params,
    epocas=EPOCAS_FINAL,
    verbose=True
)

# Avaliação final no conjunto de teste independente
teste_src_device = teste_src.to(DEVICE)
teste_tgt_device = teste_tgt.to(DEVICE)

pred = modelo.prever(teste_src_device, passos=LEN_SAIDA)

mse_modelo = nn.functional.mse_loss(pred, teste_tgt_device).item()

# Baseline: repete o último valor observado da série
baseline = teste_src_device[:, -1:, :].expand(-1, LEN_SAIDA, -1)
mse_base = nn.functional.mse_loss(baseline, teste_tgt_device).item()
reducao = (1 - mse_modelo / mse_base) * 100

print("\n========== RESULTADO FINAL ==========")
print("Melhor trial:", estudo.best_trial.number)
print("Melhor época:", modelo.melhor_epoca)
print(f"Melhor loss na validação: {modelo.melhor_loss_auto:.5f}")
print(f"MSE Transformer no teste: {mse_modelo:.5f}")
print(f"MSE baseline: {mse_base:.5f}")
print(f"Redução do erro: {reducao:.2f}%")

# ---------- Gráficos comparativos das três variáveis no teste ----------
exemplo = 0
nomes = ["Seno", "Cosseno", "Onda mista"]

passado = teste_src_device[exemplo].cpu().numpy()
futuro_real = teste_tgt_device[exemplo].cpu().numpy()
futuro_pred = pred[exemplo].cpu().numpy()

x_passado = np.arange(LEN_ENTRADA)
x_futuro = np.arange(LEN_ENTRADA, LEN_ENTRADA + LEN_SAIDA)

fig, axes = plt.subplots(3, 1, figsize=(12, 9), sharex=True)

for var, ax in enumerate(axes):
    ax.plot(x_passado, passado[:, var], label="Passado (entrada)")
    ax.plot(x_futuro, futuro_real[:, var], "g-o", label="Futuro real")
    ax.plot(x_futuro, futuro_pred[:, var], "r--o", label="Previsão Transformer")
    ax.set_title(nomes[var])
    ax.set_ylabel("Valor normalizado")
    ax.grid(alpha=0.3)
    ax.legend()

axes[-1].set_xlabel("Passos no tempo")
plt.suptitle("Previsão do Transformer nas três séries (Conjunto de Teste)", fontsize=14)
plt.tight_layout()
plt.show()
```
